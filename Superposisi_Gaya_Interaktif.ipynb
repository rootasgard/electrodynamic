# Komputasi Superposisi Gaya Listrik (Interaktif)
Notebook ini menggunakan pustaka `plotly` agar plot 3D dapat dirotasi, di-zoom, dan digeser secara interaktif, mirip dengan antarmuka di aplikasi seperti Maple.

```python
# Pastikan Anda sudah menginstal plotly: pip install plotly
import numpy as np
import plotly.graph_objects as go

def hitung_superposisi_gaya():
    # Konstanta Coulomb (N m^2 / C^2)
    k = 9e9
    
    # Muatan uji q di titik asal (0,0,0) dalam Coulomb
    q = 10e-6 
    r_q = np.array([0.0, 0.0, 0.0])
    
    # Data muatan sumber (q_i dalam Coulomb, posisi dalam meter)
    muatan_sumber = [
        {'q': 2e-6,  'pos': np.array([0.0, 2.0, 0.0]), 'label': 'q1 (2 µC)'},
        {'q': 4e-6,  'pos': np.array([2.0, 2.0, 0.0]), 'label': 'q2 (4 µC)'},
        {'q': 5e-6,  'pos': np.array([0.0, 2.0, 2.0]), 'label': 'q3 (5 µC)'},
        {'q': -5e-6, 'pos': np.array([2.0, 2.0, 2.0]), 'label': 'q4 (-5 µC)'}
    ]
    
    # Vektor gaya total
    F_total = np.zeros(3)
    
    print("--- Perhitungan Gaya Superposisi ---")
    for i, muatan in enumerate(muatan_sumber):
        r_i = muatan['pos']
        q_i = muatan['q']
        
        # Vektor jarak dari sumber ke muatan uji
        r_vec = r_q - r_i
        r_mag = np.linalg.norm(r_vec)
        
        # Menghitung gaya Coulomb
        F_i = k * q * q_i * (r_vec / (r_mag**3))
        F_total += F_i
        
        print(f"Gaya dari {muatan['label']} terhadap q: {F_i} N")
        
    print("-" * 40)
    print(f"Vektor Gaya Total (F_total): {F_total} N")
    print(f"Besar Gaya Total (|F_total|): {np.linalg.norm(F_total):.4f} N")
    
    return muatan_sumber, r_q, F_total

def plot_model_3d_interaktif(muatan_sumber, r_q, F_total):
    fig = go.Figure()
    
    # Plot muatan uji q
    fig.add_trace(go.Scatter3d(
        x=[r_q[0]], y=[r_q[1]], z=[r_q[2]],
        mode='markers+text',
        marker=dict(size=8, color='red'),
        text=['q (10 µC, Uji)'],
        textposition='top center',
        name='Muatan Uji'
    ))
    
    # Plot muatan sumber
    for muatan in muatan_sumber:
        pos = muatan['pos']
        warna = 'blue' if muatan['q'] > 0 else 'green'
        fig.add_trace(go.Scatter3d(
            x=[pos[0]], y=[pos[1]], z=[pos[2]],
            mode='markers+text',
            marker=dict(size=6, color=warna),
            text=[muatan['label']],
            textposition='bottom center',
            name=muatan['label']
        ))
        
    # Plot Vektor Gaya Total (F_total)
    skala_visual = 1.0 / (np.linalg.norm(F_total) + 1e-9)
    v_end = r_q + (F_total * skala_visual * 2.5) # Mengatur panjang garis panah
    
    # Membuat garis panah (vektor)
    fig.add_trace(go.Scatter3d(
        x=[r_q[0], v_end[0]],
        y=[r_q[1], v_end[1]],
        z=[r_q[2], v_end[2]],
        mode='lines',
        line=dict(color='red', width=5),
        name='Garis F_total'
    ))
    
    # Menambahkan ujung panah (cone)
    fig.add_trace(go.Cone(
        x=[v_end[0]], y=[v_end[1]], z=[v_end[2]],
        u=[F_total[0]], v=[F_total[1]], w=[F_total[2]],
        sizemode="absolute", sizeref=0.4,
        showscale=False,
        colorscale=[[0, 'red'], [1, 'red']],
        name='Arah F_total'
    ))
    
    # Konfigurasi Layout agar proporsional dan interaktif
    fig.update_layout(
        title="Plot 3D Superposisi Gaya Listrik (Interaktif)",
        scene=dict(
            xaxis_title='Sumbu X (m)',
            yaxis_title='Sumbu Y (m)',
            zaxis_title='Sumbu Z (m)',
            xaxis=dict(range=[-1, 3]),
            yaxis=dict(range=[-1, 3]),
            zaxis=dict(range=[-1, 3]),
            aspectmode='cube' # Menjaga rasio kubus agar tidak distorsi
        ),
        margin=dict(l=0, r=0, b=0, t=40),
        showlegend=True
    )
    
    fig.show()

muatan_sumber, r_q, F_total = hitung_superposisi_gaya()
plot_model_3d_interaktif(muatan_sumber, r_q, F_total)
```
