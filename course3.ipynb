## 2. Gaya Lorentz

Gaya pada muatan uji $q$ yang bergerak dengan kecepatan $\mathbf v$ dalam medan listrik dan magnet adalah

$$\mathbf F=q(\mathbf E+\mathbf v\times\mathbf B),\qquad \mathbf F_{\rm mag}=q\,\mathbf v\times\mathbf B.$$

Besar gaya magnet $|q|vB\sin\theta$ hanya bergantung pada komponen kecepatan yang tegak lurus $\mathbf B$: $v_\perp=v\sin\theta$. Komponen sejajar medan tidak menghasilkan gaya. Arah $\mathbf v\times\mathbf B$ berlaku untuk $q>0$; muatan negatif membalik arah gaya. Karena $\mathbf F_{\rm mag}\cdot\mathbf v=0$, gaya magnet sesaat tidak mengubah laju/energi kinetik partikel, meski dapat membelokkan lintasannya. (Griffiths §5.1.2, hlm. 212–216.)

### Penurunan komponen dan pembacaan arah pada diagram

Hukum Lorentz adalah **hukum dasar empiris**: bentuk perkalian silang menyatakan hasil eksperimen dan menyatakan besar serta arah gaya. Dengan vektor masukan $\mathbf v=(v_x,v_y,v_z)$ dan $\mathbf B=(B_x,B_y,B_z)$, hitung komponen hasil sebagai

$$
\mathbf v\times\mathbf B=
\begin{vmatrix}
\hat{\mathbf x}&\hat{\mathbf y}&\hat{\mathbf z}\\
v_x&v_y&v_z\\
B_x&B_y&B_z
\end{vmatrix}
=(v_yB_z-v_zB_y)\hat{\mathbf x}+(v_zB_x-v_xB_z)\hat{\mathbf y}+(v_xB_y-v_yB_x)\hat{\mathbf z}.
$$

**Cocokkan masukan dan hasil dengan gambar di atas.** Sumbu $x$ ke kanan, $y$ ke atas, dan $+z$ keluar bidang (simbol titik). Gambar memasukkan $\mathbf v=v\hat{\mathbf x}$ dan $\mathbf B=B\hat{\mathbf z}$, sehingga

$$\mathbf v\times\mathbf B=vB(\hat{\mathbf x}\times\hat{\mathbf z})=-vB\hat{\mathbf y},\qquad \mathbf F_{\rm mag}=-qvB\hat{\mathbf y}\quad(q>0).$$

Jadi panah hasil gaya untuk $q>0$ memang ke bawah; untuk $q<0$, hasil dikalikan muatan negatif dan arahnya ke atas. Jika ada komponen kecepatan sejajar $\mathbf B$, komponen itu tidak muncul dalam determinan setelah cross product. Besarnya dapat dibaca lewat proyeksi $v_\perp=v\sin\theta$: $|\mathbf F_{\rm mag}|=|q|Bv_\perp$.

### Proyeksi kecepatan dan besar gaya

Jika $\mathbf B=B\hat{\mathbf z}$ dan $\mathbf v=v_x\hat{\mathbf x}+v_y\hat{\mathbf y}+v_z\hat{\mathbf z}$, maka

$$\mathbf v\times\mathbf B=B(v_y\hat{\mathbf x}-v_x\hat{\mathbf y}),\qquad F_{\rm mag}=|q|B\sqrt{v_x^2+v_y^2}.$$

Komponen $v_z$ sejajar medan hilang dari hasil silang. Secara umum, proyeksikan $\mathbf v$ menjadi $\mathbf v_\parallel+\mathbf v_\perp$; hanya $\mathbf v_\perp$ yang menentukan gaya. Ini juga menjelaskan mengapa gerak dalam medan seragam dapat berupa heliks: gerak melingkar dari komponen tegak lurus ditambah gerak lurus dari komponen sejajar.

## 3. Arus dan gaya pada pembawa arus

Arus konvensional berarah seperti aliran muatan positif. Untuk kawat, $I$ adalah muatan yang melintasi penampang per satuan waktu; pada rumus elemen kawat, $I$ biasanya besar arus dan $d\boldsymbol\ell$ dipilih searah arus. Rapat arus volume $\mathbf J$ (A/m$^2$) dan rapat arus permukaan $\mathbf K$ (A/m) adalah vektor.

Penjumlahan gaya Lorentz atas muatan bergerak menghasilkan bentuk kontinu

$$d\mathbf F=I\,d\boldsymbol\ell\times\mathbf B,\qquad \mathbf f=\mathbf J\times\mathbf B.$$

Di sini $d\boldsymbol\ell$ berarah sama dengan arus konvensional dan $\mathbf f$ adalah gaya per volume. Untuk arus tunak, tidak ada akumulasi muatan lokal: persamaan kontinuitas $\partial\rho/\partial t+\nabla\cdot\mathbf J=0$ menjadi $\nabla\cdot\mathbf J=0$. (Griffiths §5.1.3 dan §5.2.1, hlm. 216–224.)

### Dari gaya satu muatan ke gaya pada kawat

Ambil segmen lurus kecil dengan rapat muatan bergerak linear $\lambda$ dan kecepatan pembawa $\mathbf v=v\hat{\mathbf t}$ searah arus konvensional. Muatan pada panjang $d\ell$ adalah $dq=\lambda d\ell$, sedangkan $I=\lambda v$. Karena itu

$$d\mathbf F=(dq\,\mathbf v)\times\mathbf B=(\lambda d\ell\,v\hat{\mathbf t})\times\mathbf B=I\,d\boldsymbol\ell\times\mathbf B.$$

Urutan masukan perkalian silang ialah **elemen arus dahulu, medan kemudian**. Untuk medan konstan, integralkan sepanjang kawat: $\mathbf F=I\int d\boldsymbol\ell\times\mathbf B$. Untuk arus volume, muatan bergerak per volume menghasilkan gaya per volume $\mathbf f=\mathbf J\times\mathbf B$ (N/m$^3$), sehingga $d\mathbf F=\mathbf f\,d\tau$.

## 4. Hukum Biot–Savart

Untuk elemen kawat berarus tunak dan untuk distribusi arus volume,

$$d\mathbf B(\mathbf r)=\frac{\mu_0}{4\pi}\frac{I\,d\boldsymbol\ell'\times\hat{\mathbf R}}{R^2},\qquad
\mathbf B(\mathbf r)=\frac{\mu_0}{4\pi}\int\frac{\mathbf J(\mathbf r')\times\hat{\mathbf R}}{R^2}\,d\tau'.$$

Arah $d\mathbf B$ tegak lurus bidang yang dibentuk $d\boldsymbol\ell'$ dan $\mathbf R$, bukan sepanjang garis sumber–pengamat. Besar elemennya sebanding $\sin\theta$, sehingga kontribusi nol jika kedua vektor sejajar. Integrasi menjumlahkan kontribusi sebagai **vektor**. (Griffiths §5.2.2, hlm. 224–228.)

### Membaca vektor sumber dan medan pada sketsa

Untuk elemen garis, masukannya adalah arus skalar positif $I$, elemen vektor $d\boldsymbol\ell'$ **searah arus**, dan vektor pemisah $\mathbf R=\mathbf r-\mathbf r'$ **dari elemen sumber ke titik medan**. Hasil $d\mathbf B$ tegak lurus kedua vektor masukan:

$$d\mathbf B=\frac{\mu_0 I}{4\pi}\frac{d\boldsymbol\ell'\times\hat{\mathbf R}}{R^2}=\frac{\mu_0 I}{4\pi}\frac{d\boldsymbol\ell'\times\mathbf R}{R^3}.$$

Pada diagram di bawah, kawat dan $d\boldsymbol\ell'$ menunjuk $+z$, titik medan ada di $+x$, maka bagian $+x$ pada $\mathbf R$ memberi $\hat{\mathbf z}\times\hat{\mathbf x}=+\hat{\mathbf y}$. Komponen $\mathbf R$ sejajar $z$ tidak berkontribusi karena $\hat{\mathbf z}\times\hat{\mathbf z}=0$; jadi simbol titik (keluar bidang) untuk $d\mathbf B$ konsisten dengan masukan gambar. Membalik arus membalik arah hasil. Untuk distribusi volume, jumlahkan elemen sumber $\mathbf J(\mathbf r')d\tau'$:

$$\mathbf B(\mathbf r)=\frac{\mu_0}{4\pi}\int\frac{\mathbf J(\mathbf r')\times\hat{\mathbf R}}{R^2}\,d\tau'.$$


### Kawat Lurus Tak Hingga

Ambil kawat sepanjang sumbu $z$ dengan arus $+z$, dan titik pengamatan berjarak tegak lurus $s$. Simetri silinder memaksa medan melingkar mengitari kawat. Dalam koordinat silinder, $\hat{\boldsymbol\phi}=\hat{\mathbf z}\times\hat{\mathbf s}$; integral Biot–Savart memberi

$$\mathbf B(s)=\frac{\mu_0 I}{2\pi s}\,\hat{\boldsymbol\phi}.$$

Untuk melihat proyeksi, pada titik sumbu $+x$ arah $\hat{\boldsymbol\phi}=+\hat y$. Setiap elemen kawat memberi arah tangensial ini; besar kontribusinya berubah sepanjang kawat tetapi arahnya sama pada sumbu tersebut. Membalik arus membalik $\mathbf B$.

#### Turunan dari elemen sumber sampai medan total

Pilih sumber pada kawat sebagai $\mathbf r'=z'\hat{\mathbf z}$ dan titik pengamatan $\mathbf r=s\hat{\mathbf x}$. Dengan arus $I$ ke $+z$,

$$d\boldsymbol\ell'=dz'\hat{\mathbf z},\quad \mathbf R=s\hat{\mathbf x}-z'\hat{\mathbf z},\quad R=\sqrt{s^2+z'^2}.$$

Substitusi **masukan vektor** ke Biot–Savart (bentuk $\mathbf R/R^3$):

$$d\mathbf B=\frac{\mu_0I}{4\pi}\frac{dz'\hat{\mathbf z}\times(s\hat{\mathbf x}-z'\hat{\mathbf z})}{(s^2+z'^2)^{3/2}}=\frac{\mu_0I}{4\pi}\frac{s\,dz'}{(s^2+z'^2)^{3/2}}\hat{\mathbf y}.$$

Semua elemen memiliki arah hasil $+y$, sehingga integral tinggal menjumlahkan besar kontribusinya. Dengan $u=z'/s$,

$$\mathbf B=\frac{\mu_0I}{4\pi}\hat{\mathbf y}\int_{-\infty}^{\infty}\frac{s\,dz'}{(s^2+z'^2)^{3/2}}=\frac{\mu_0I}{4\pi s}\hat{\mathbf y}\int_{-\infty}^{\infty}\frac{du}{(1+u^2)^{3/2}}=\frac{\mu_0I}{2\pi s}\hat{\mathbf y}.$$

Ini adalah medan pada sumbu $+x$; bentuk di titik umum silinder ialah $\mathbf B=(\mu_0I/2\pi s)\hat{\boldsymbol\phi}$. Nilai $\hat{\boldsymbol\phi}=+\hat{\mathbf y}$ tepat di $+x$, sesuai sketsa arah medan.

### Kawat Melingkar / Loop Lingkaran (proyeksi medan ke sumbu)

Pada titik di sumbu loop berjari-jari $R$, jarak dari setiap elemen ke titik pengamatan sama, $r=\sqrt{R^2+z^2}$. Pasangan elemen yang berhadapan menghasilkan komponen radial yang sama besar tetapi berlawanan arah; komponen radial **saling meniadakan**. Komponen sumbu $z$ searah dan dijumlahkan. Hasilnya

$$\mathbf B(z)=\frac{\mu_0 I R^2}{2(R^2+z^2)^{3/2}}\,\hat{\mathbf z}.$$

Ini contoh penting: proyeksi bukan sekadar menggambar panah, tetapi memilih komponen yang bertahan setelah integrasi simetris. (Griffiths §5.2.2, contoh loop pada hlm. 226–227.)

### Turunan loop: Proyeksikan sebelum mengintegralkan

Letakkan loop berjari-jari $a$ di bidang $xy$, pusat di asal, arus berlawanan arah jarum jam jika dilihat dari $+z$, dan pengamat di $\mathbf r=z\hat{\mathbf z}$. Elemen sumber pada sudut $\phi$ dan masukan vektornya adalah

$$\mathbf r'=a\hat{\boldsymbol\rho}',\quad d\boldsymbol\ell'=a\,d\phi\,\hat{\boldsymbol\phi}',\quad \mathbf R=-a\hat{\boldsymbol\rho}'+z\hat{\mathbf z},\quad R=\sqrt{a^2+z^2}.$$

Dari identitas basis silinder $\hat{\boldsymbol\phi}'\times\hat{\boldsymbol\rho}'=-\hat{\mathbf z}$ dan $\hat{\boldsymbol\phi}'\times\hat{\mathbf z}=\hat{\boldsymbol\rho}'$:

$$d\boldsymbol\ell'\times\mathbf R=a\,d\phi\left(a\hat{\mathbf z}+z\hat{\boldsymbol\rho}'\right),\quad d\mathbf B=\frac{\mu_0I a\,d\phi}{4\pi(a^2+z^2)^{3/2}}\left(a\hat{\mathbf z}+z\hat{\boldsymbol\rho}'\right).$$

Komponen hasil radial $z\hat{\boldsymbol\rho}'$ saling meniadakan untuk elemen berseberangan ($\phi$ dan $\phi+\pi$), sedangkan proyeksi $+z$ sama arah untuk seluruh loop. Maka

$$\mathbf B(z)=\frac{\mu_0I a^2}{4\pi(a^2+z^2)^{3/2}}\int_0^{2\pi}d\phi\,\hat{\mathbf z}=\frac{\mu_0I a^2}{2(a^2+z^2)^{3/2}}\hat{\mathbf z}.$$

**Orientasi sketsa proyeksi:** bidang gambar adalah $xz$; sumbu $y$ tegak lurus kertas. Pada titik kawat $+x$, arus CCW mengarah $+y$ (titik/keluar kertas); di $-x$, arus mengarah $-y$ (silang/masuk kertas). Panah kontribusi di titik $P$ karenanya memiliki proyeksi radial berlawanan, tetapi komponen vertikal $+z$ sama.


```python
# Loop arus berada di bidang xy, dengan titik pengamatan pada sumbu z.
a, zP = 1.6, 2.2
loop3d = []
for i in range(129):
    phi = 2*np.pi*i/128
    loop3d.append([a*np.cos(phi), a*np.sin(phi), 0])
Rplus = np.array([0.0, 0.0, zP]) - np.array([a, 0.0, 0.0])
Rminus = np.array([0.0, 0.0, zP]) - np.array([-a, 0.0, 0.0])
tampilkan_scene_3d({
    "title": "Simetri loop arus melingkar",
    "description": "Untuk elemen yang berseberangan, komponen radial dB saling meniadakan, sedangkan komponen aksialnya dijumlahkan.",
    "axes": {"length": 3.8},
    "lines": [{"points": loop3d, "color": "#b87333", "radius": 0.028}],
    "points": [{"position": [a,0,0], "label": "S+", "color": "#c2413b"}, {"position": [-a,0,0], "label": "S−", "color": "#c2413b"}, {"position": [0,0,zP], "label": "P", "color": "#202d32"}],
    "arrows": [
        {"from": [a,0,0], "to": [a,0.45,0], "color": "#2563eb", "label": "dℓ′+"},
        {"from": [-a,0,0], "to": [-a,-0.45,0], "color": "#2563eb", "label": "dℓ′−"},
        {"from": [a,0,0], "to": [0,0,zP], "color": "#b86b16", "label": "R+"},
        {"from": [-a,0,0], "to": [0,0,zP], "color": "#16836e", "label": "R−"},
        {"from": [0,0,zP], "to": [0.55,0,zP+0.45], "color": "#7b4fa3", "label": "dB+"},
        {"from": [0,0,zP], "to": [-0.55,0,zP+0.45], "color": "#7b4fa3", "label": "dB−"},
        {"from": [0,0,zP], "to": [0,0,zP+0.9], "color": "#c2413b", "label": "dBz"}
    ],
    "legend": ["Arus loop berarah berlawanan arah jarum jam jika dilihat dari +z.", "Komponen radial kontribusi berlawanan; komponen z sama-sama positif."],
    "view": {"camera": [5, 5, 7], "target": [0,0,1.2]}
})
```

### Solenoida panjang: penjumlahan medan loop dengan Biot–Savart

Anggap solenoida berjari-jari $a$ memiliki $n=N/L$ lilitan per satuan panjang dan arus $I$. Sumbu solenoida adalah $z$; titik pengamatan berada pada sumbu di koordinat $z$. Turunan dimulai dari medan satu loop yang baru diperoleh:

$$dB_{\mathrm{loop},z}=\frac{\mu_0 I a^2}{2\left[a^2+(z-z')^2\right]^{3/2}}.$$

Pada elemen panjang $dz'$ terdapat $dN=n\,dz'$ loop. Karena loop-loop koaksial, komponen transversal tiap loop saling meniadakan oleh simetri putar; hanya komponen $z$ yang dijumlahkan. Maka

$$B_z(z)=\int dN\,dB_{\mathrm{loop},z}=\frac{\mu_0 nI a^2}{2}\int_{z_1}^{z_2}\frac{dz'}{\left[a^2+(z-z')^2\right]^{3/2}}.$$

Gunakan $u=z-z'$, sehingga $du=-dz'$ dan antiturunan $\int du/(a^2+u^2)^{3/2}=u/[a^2\sqrt{a^2+u^2}]$. Setelah memasukkan batas, diperoleh

$$\mathbf B(z)=\frac{\mu_0 nI}{2}\left[\frac{z-z_1}{\sqrt{a^2+(z-z_1)^2}}-\frac{z-z_2}{\sqrt{a^2+(z-z_2)^2}}\right]\hat{\mathbf z}.$$

Untuk solenoida sangat panjang dan titik jauh dari kedua ujung, suku pertama mendekati $+1$ dan suku kedua $-1$, sehingga $\mathbf B\simeq\mu_0nI\hat{\mathbf z}$ di dalam. Di luar solenoida ideal tak hingga, kontribusi ujung meniadakan medan; solenoida berhingga memiliki medan tepi dan medan luar kecil. Tanda medan dibalik jika arah arus dibalik.

**Proyeksi vektor.** Untuk sebuah elemen kawat pada radius $a$, $d\boldsymbol\ell'$ tangensial terhadap loop dan $\mathbf R$ mengarah dari elemen ke $P$. Perkalian silang memberi komponen aksial dan radial. Elemen yang berhadapan menghapus komponen radial; proyeksi aksial searah sepanjang semua loop. Gambar berikut menampilkan pasangan proyeksi itu.

```python
# Solenoida berhingga dibentuk oleh loop-loop pada posisi z yang berbeda.
zturns = np.linspace(-3.2, 3.2, 11)
turns = []
for zc in zturns:
    pts = [[np.cos(t), np.sin(t), float(zc)] for t in np.linspace(0, 2*np.pi, 97)]
    turns.append({"points": pts, "color": "#b87333", "radius": 0.026})
tampilkan_scene_3d({
    "title": "Medan di dalam solenoida",
    "description": "Loop-loop koaksial menghasilkan komponen medan aksial searah di bagian dalam solenoida.",
    "axes": {"length": 4.0}, "lines": turns,
    "arrows": [
        {"from": [-2.6,0,0], "to": [-1.3,0,0], "color": "#b86b16", "label": "R+"},
        {"from": [-2.6,0,0], "to": [-1.3,0,0.45], "color": "#b86b16", "label": "R−"},
        {"from": [-1.2,0,0.2], "to": [-0.5,0,0.75], "color": "#7b4fa3", "label": "dB+"},
        {"from": [-1.2,0,-0.2], "to": [-0.5,0,-0.75], "color": "#7b4fa3", "label": "dB−"},
        {"from": [-1.7,0.55,1.2], "to": [1.7,0.55,1.2], "color": "#c2413b", "label": "B ≈ μ₀nI ẑ"}
    ],
    "legend": ["Panah ungu: pasangan kontribusi; komponen radial saling meniadakan.", "Panah merah: medan total mengarah sepanjang sumbu solenoida."],
    "view": {"camera": [7, 5.5, 8]}
})
```

### Toroida: medan dari superposisi Biot–Savart

Toroida ideal adalah solenoida panjang yang dilengkungkan menjadi cincin berjari-jari dalam $b$ dan luar $c$, dengan $N$ lilitan berarus $I$. Untuk elemen kawat pada lilitan ke-$k$, Biot–Savart memberi

$$d\mathbf B_k(\mathbf r)=\frac{\mu_0 I}{4\pi}\frac{d\boldsymbol\ell'_k\times(\mathbf r-\mathbf r'_k)}{|\mathbf r-\mathbf r'_k|^3}.$$

Medan total diperoleh dengan menjumlahkan integral atas semua lilitan:

$$\mathbf B(\mathbf r)=\frac{\mu_0 I}{4\pi}\sum_{k=1}^{N}\oint_{C_k}\frac{d\boldsymbol\ell'_k\times(\mathbf r-\mathbf r'_k)}{|\mathbf r-\mathbf r'_k|^3}.$$

**Langkah proyeksi.** Evaluasi perkalian silang untuk tiap elemen sumber, lalu uraikan $d\mathbf B_k$ ke komponen radial, azimutal, dan aksial terhadap sumbu toroida. Untuk lilitan yang terdistribusi rapat dan seragam, pasangan elemen simetris meniadakan komponen radial dan aksial. Yang bertahan adalah $B_\phi\hat{\boldsymbol\phi}$. Karena lilitan nyata bersifat diskret, jumlah integral Biot–Savart di atas merupakan rumus umum; menghitung jumlah itu secara langsung untuk lilitan tertentu biasanya memerlukan penjumlahan numerik.

**Menentukan besar medan ideal tahap demi tahap.** Simetri toroida membuat $B_\phi$ sama besar pada lingkaran berjari-jari $\rho$ yang koaksial dengan toroid dan arahnya sejajar $d\boldsymbol\ell$. Pilih permukaan lingkaran itu sebagai lintasan Amperian:

$$\oint_C\mathbf B\cdot d\boldsymbol\ell=B_\phi(\rho)\,2\pi\rho=\mu_0 I_{\rm enc}.$$

Untuk $b<\rho<c$, permukaan yang dibatasi lintasan memotong semua $N$ lilitan satu kali, sehingga $I_{\rm enc}=NI$ dan

$$\mathbf B(\rho)=\frac{\mu_0NI}{2\pi\rho}\,\hat{\boldsymbol\phi},\qquad b<\rho<c.$$

Untuk toroida ideal dengan lilitan kontinu, arus tertutup bersih nol jika $\rho<b$ (lubang tengah) atau $\rho>c$ (arus masuk-keluar permukaan saling meniadakan), sehingga $\mathbf B=0$ pada kedua wilayah itu. Lilitan yang terpisah dan inti berhingga menimbulkan medan bocor. Di inti magnetik linear dan homogen, pendekatan ideal memberi $\mathbf B=\mu\mathbf H$ dengan $H_\phi=NI/(2\pi\rho)$; untuk bahan nonlinier, permeabilitas tidak boleh dianggap konstan.

$$\mathbf B(\rho)\simeq\frac{\mu_0NI}{2\pi\rho}\,\hat{\boldsymbol\phi},$$

dengan $\rho$ jarak titik dari sumbu simetri toroida. Medan lebih besar di sisi dalam cincin karena berbanding terbalik dengan $\rho$. Arah $\hat{\boldsymbol\phi}$ dibalik jika arah arus lilitan dibalik.

**Makna Biot–Savart di sini:** integral vektor menunjukkan sumber medan dan arah proyeksi; simetri menjamin bentuk azimutal. Hukum Ampère dipakai untuk menilai besar medan ideal secara tertutup. Kedua hukum magnetostatik konsisten, tetapi turunan skalar $1/\rho$ dari integral Biot–Savart tidak sesederhana kasus kawat lurus atau loop.

```python
# Toroida ideal, garis Amperian, dan arah medan azimutal.
b, c, rhoP = 1.35, 2.45, 1.9
turns = []
for ang in np.linspace(0, 2*np.pi, 12, endpoint=False):
    pts = []
    for t in np.linspace(0, 2*np.pi, 81):
        pts.append([(b+c)*0.5*np.cos(ang) + 0.42*np.cos(t)*np.cos(ang), (b+c)*0.5*np.sin(ang) + 0.42*np.cos(t)*np.sin(ang), 0.42*np.sin(t)])
    turns.append({"points": pts, "color": "#b87333", "radius": 0.026})
tampilkan_scene_3d({
    "title": "Toroida dan lintasan Amperian",
    "description": "Medan azimutal berada di inti toroid; garis Amperian koaksial memiliki besar medan konstan pada model ideal.",
    "axes": {"length": 3.3},
    "tori": [{"radius": (b+c)/2, "tube": (c-b)/2, "color": "#d9a441", "opacity": 0.35}],
    "lines": turns,
    "circles": [{"center": [0,0,0], "radius": rhoP, "axis": "z", "color": "#7b4fa3", "dashed": True}],
    "points": [{"position": [rhoP,0,0], "label": "P(ρ)", "color": "#202d32"}],
    "arrows": [
        {"from": [1.65,0,0], "to": [1.65,0.8,0], "color": "#c2413b", "label": "Bφ"},
        {"from": [2.0,0.1,0], "to": [2.0,0.55,0], "color": "#16836e", "label": "dℓ′"},
        {"from": [2.0,0.1,0], "to": [rhoP,0,0], "color": "#b86b16", "label": "R"}
    ],
    "legend": ["Kuning transparan: inti toroida; biru: lilitan arus.", "Garis ungu putus-putus: lintasan Amperian."],
    "view": {"camera": [5.5, 5, 7]}
})
```

## 5. Divergensi, Curl, dan Hukum Ampère

Dalam magnetostatika, hukum lokal medan adalah

$$\nabla\cdot\mathbf B=0,\qquad \nabla\times\mathbf B=\mu_0\mathbf J.$$

Divergensi nol menyatakan tidak ada sumber/tenggelam titik bagi garis $\mathbf B$ (tidak ada monopole magnet dalam model ini); garis medan tidak berawal atau berakhir pada muatan magnet. Curl menyatakan arus tunak merupakan sumber sirkulasi medan. Dengan teorema Stokes, bentuk kedua menjadi

$$\oint_C\mathbf B\cdot d\boldsymbol\ell=\mu_0 I_{\rm enc},\qquad I_{\rm enc}=\int_S\mathbf J\cdot d\mathbf a.$$

Arah normal $d\mathbf a$ dan arah keliling $C$ harus konsisten aturan tangan kanan: ibu jari sepanjang normal positif, jari melengkung sepanjang lintasan positif. Ampère sangat efisien jika simetri membuat $\mathbf B$ konstan dan sejajar/tegak lurus di bagian lintasan yang dipilih; jika tidak, gunakan Biot–Savart. Hukum lokal ini berlaku untuk arus tunak, bukan bentuk lengkap untuk medan berubah waktu. (Griffiths §5.3, hlm. 229–242.)

### Turunan bentuk integral dan orientasi normal

Mulai dari persamaan lokal $\nabla\times\mathbf B=\mu_0\mathbf J$. Integralkan pada permukaan $S$ ber-normal positif $d\mathbf a$ dan terapkan teorema Stokes:

$$\int_S(\nabla\times\mathbf B)\cdot d\mathbf a=\oint_{\partial S}\mathbf B\cdot d\boldsymbol\ell=\mu_0\int_S\mathbf J\cdot d\mathbf a=\mu_0I_{\rm enc}.$$

Aturan orientasi pada gambar: ibu jari menuju normal $+z$ (arus terhitung positif), jari melengkung berlawanan arah jarum jam menentukan arah lintasan positif. Untuk kawat lurus pada sumbu $z$ dan lintasan lingkaran radius $s$, simetri memberi $\mathbf B=B_\phi\hat{\boldsymbol\phi}$ konstan; karena sejajar lintasan,

$$B_\phi\oint d\ell=B_\phi(2\pi s)=\mu_0I\quad\Rightarrow\quad\mathbf B=\frac{\mu_0I}{2\pi s}\hat{\boldsymbol\phi}.$$

Jika lintasan tidak memiliki simetri yang membuat $B$ konstan/sejajar, persamaan Ampère tetap benar tetapi tidak cukup sendirian untuk menentukan $\mathbf B$ dengan mudah.


```python
R, Iloop = 1.0, 1.0
z = np.linspace(-3, 3, 400)
Bz = mu0*Iloop*R**2/(2*(R**2+z**2)**1.5)
loop_points = [[R*np.cos(t), R*np.sin(t), 0] for t in np.linspace(0, 2*np.pi, 129)]
graph_points = [[float(3.5+zi), float(1.6*bi/Bz.max()), -1.8] for zi,bi in zip(z,Bz)]
graph_axes = [
    {"points": [[0,0,-1.8],[0,1.9,-1.8]], "color": "#495b62"},
    {"points": [[0,0,-1.8],[7,0,-1.8]], "color": "#495b62"}
]
tampilkan_scene_3d({
    "title": "Loop arus dan profil medan aksial",
    "description": "Panel kanan adalah kurva Bz(z) dalam ruang 3D; ordinat dinormalisasi terhadap nilai maksimumnya di z = 0.",
    "axes": {"length": 1.6},
    "lines": [{"points": loop_points,"color":"#2563eb"}, *graph_axes],
    "curves": [{"points": graph_points,"color":"#c2413b","radius":.025}],
    "points": [{"position":[0,0,0],"label":"I_loop","color":"#c2413b"}],
    "legend": ["Loop berada pada bidang xy.", "Sumbu grafik: z/a (horizontal), Bz/Bz(0) (vertikal)."],
    "view": {"camera":[7,7,11],"target":[3.7,.7,-.6]}
})
```

## 6. Vektor Potensial Magnet

Karena $\nabla\cdot\mathbf B=0$, medan magnet dapat ditulis sebagai curl suatu medan vektor:

$$\mathbf B=\nabla\times\mathbf A.$$

Identitas $\nabla\cdot(\nabla\times\mathbf A)=0$ membuat syarat tanpa monopole otomatis terpenuhi. Potensial vektor tidak tunggal: untuk fungsi skalar $\chi$, transformasi

$$\mathbf A'=\mathbf A+\nabla\chi$$

tidak mengubah medan karena $\nabla\times\nabla\chi=0$. Kebebasan ini disebut kebebasan gauge. Pilihan Coulomb $\nabla\cdot\mathbf A=0$ menyederhanakan hukum Ampère menjadi persamaan Poisson (komponen Kartesius):

$$\nabla^2\mathbf A=-\mu_0\mathbf J,\qquad
\mathbf A(\mathbf r)=\frac{\mu_0}{4\pi}\int\frac{\mathbf J(\mathbf r')}{|\mathbf r-\mathbf r'|}\,d\tau'.$$

Untuk arus garis, $\mathbf A(\mathbf r)=\frac{\mu_0 I}{4\pi}\int d\boldsymbol\ell'/R$. Medan yang terukur adalah $\mathbf B$; $\mathbf A$ adalah representasi yang bergantung gauge dan tidak boleh disamakan dengan $\mathbf B$. (Griffiths §5.4.1, hlm. 243–249.)

### Turunan persamaan Poisson untuk $\mathbf A$

Mulai dari definisi $\mathbf B=\nabla\times\mathbf A$ dan identitas curl-curl:

$$\nabla\times\mathbf B=\nabla\times(\nabla\times\mathbf A)=\nabla(\nabla\cdot\mathbf A)-\nabla^2\mathbf A.$$

Kebebasan gauge $\mathbf A'=\mathbf A+\nabla\chi$ membolehkan gauge Coulomb $\nabla\cdot\mathbf A=0$. Persamaan Ampère magnetostatik lalu menjadi

$$\nabla\times\mathbf B=\mu_0\mathbf J\quad\Rightarrow\quad-\nabla^2\mathbf A=\mu_0\mathbf J\quad\Rightarrow\quad\nabla^2\mathbf A=-\mu_0\mathbf J.$$

Ini adalah persamaan Poisson untuk tiap komponen **Kartesius** $A_x,A_y,A_z$. Dengan syarat medan potensial meluruh sesuai solusi Green di tak hingga, inversi Poisson memberi

$$\mathbf A(\mathbf r)=\frac{\mu_0}{4\pi}\int\frac{\mathbf J(\mathbf r')}{|\mathbf r-\mathbf r'|}\,d\tau'.$$

Untuk kawat garis, gantikan $\mathbf Jd\tau'$ dengan $I\,d\boldsymbol\ell'$ sehingga $\mathbf A=(\mu_0I/4\pi)\int d\boldsymbol\ell'/R$. Setelah memperoleh $\mathbf A$, ambil rotasi (curl) (dengan turunan terhadap koordinat pengamatan yang tidak berprima) untuk mendapatkan hasil medan $\mathbf B$. Pada grafik kawat, komponen $\mathbf A=A_z(s)\hat{\mathbf z}$ memberi $B_\phi=-dA_z/ds$; nilai $A_z$ sendiri bergantung gauge.


```python
# Lintasan Ampèrian berbentuk lingkaran mengelilingi arus +z.
radius = 1.55
ampere_circle = [[radius*np.cos(t), radius*np.sin(t), 0] for t in np.linspace(0,2*np.pi,129)]
arrows = []
for t in [0.25,1.8,3.35,4.9]:
    x,y=radius*np.cos(t),radius*np.sin(t)
    arrows.append({"from":[x,y,0],"to":[x-0.35*np.sin(t),y+0.35*np.cos(t),0],"color":"#2563eb","label":"dℓ"})
arrows.extend([
    {"from":[0,0,-.2],"to":[0,0,.8],"color":"#c2413b","label":"Ienc"},
    {"from":[-.4,radius,0],"to":[.4,radius,0],"color":"#16836e","label":"Bφ"}
])
tampilkan_scene_3d({
    "title": "Orientasi lintasan pada hukum Ampère",
    "description": "Arus terlingkupi positif sepanjang +z; lintasan positif berarah berlawanan jarum jam jika dilihat dari +z.",
    "axes": {"length": 2.1}, "lines": [{"points":ampere_circle,"color":"#6b7375"}],
    "cylinders": [{"from":[0,0,-1.1],"to":[0,0,1.1],"radius":.08,"color":"#c2413b"}],
    "arrows": arrows,
    "legend": ["Hubungan integral: ∮ B·dℓ = μ₀ Ienc."],
    "view": {"camera":[4.8,5.2,7]}
})
```

```python
# Satu pilihan gauge Coulomb untuk kawat lurus tak hingga.
I, s0 = 1.0, 0.05
s = np.linspace(0.008, 0.12, 250)
Az = -(mu0*I/(2*np.pi))*np.log(s/s0)
Bphi = mu0*I/(2*np.pi*s)
cm = s*100
az_points = [[float(x),float(y*1e6),-1.0] for x,y in zip(cm,Az)]
b_points = [[float(x),float(y*1e6),1.0] for x,y in zip(cm,Bphi)]
tampilkan_scene_3d({
    "title": "Potensial vektor dan medan magnet kawat",
    "description": "Dua kurva 3D dipisahkan pada sumbu kedalaman agar skala ordinat masing-masing tetap terbaca.",
    "lines": [
        {"points":[[0,0,-1],[12,0,-1]],"color":"#495b62"},
        {"points":[[0,-.25,-1],[0,.45,-1]],"color":"#495b62"},
        {"points":[[0,0,1],[12,0,1]],"color":"#495b62"},
        {"points":[[0,0,1],[0,27,1]],"color":"#495b62"}
    ],
    "curves": [{"points":az_points,"color":"#16836e","radius":.022},{"points":b_points,"color":"#c2413b","radius":.022}],
    "legend": ["Bidang z = −1: Az(s) dalam μT·m; nilai bergantung pada gauge.", "Bidang z = +1: Bφ(s) dalam μT; Bφ = −dAz/ds."],
    "view": {"camera":[8,13,19],"target":[6,12,0]}
})
```

```python
# ==============================================================================
# Verifikasi Simbolik Formal Menggunakan SymPy
# Menghitung Medan Biot-Savart, Hukum Ampère, dan Potensial Vektor Magnetik
# ==============================================================================
import sympy as sp
z_s, R_s, I_s, s_s, mu0_s = sp.symbols('z R I s mu_0', positive=True, real=True)

print("=== VERIFIKASI SIMBOLIK FORMAL SYMPY (MAGNETOSTATIKA) ===")
# 1. Medan Biot-Savart pada sumbu simetri loop lingkaran berjejari R:
B_z_loop = (mu0_s * I_s * R_s**2) / (2 * (R_s**2 + z_s**2)**(sp.Rational(3, 2)))
display(sp.Eq(sp.Symbol('B_{z,\\text{loop}}(z)'), B_z_loop))

# Limit jarak dekat di pusat loop (z = 0):
B_z_center = B_z_loop.subs(z_s, 0)
print(f"1. Medan di pusat loop (z=0)            : B_z = {B_z_center}")

# Limit jarak jauh (aproksimasi dipol z >> R):
m_s = I_s * sp.pi * R_s**2
B_dipol_limit = sp.simplify((mu0_s * m_s) / (2 * sp.pi * z_s**3))
print(f"2. Limit dipol aksial jarak jauh (z >> R): B_z ≈ {B_dipol_limit}")

# 2. Medan magnet kawat lurus tak hingga (Hukum Ampère):
B_wire = (mu0_s * I_s) / (2 * sp.pi * s_s)
display(sp.Eq(sp.Symbol('B_{\\phi,\\text{wire}}(s)'), B_wire))
print(f"3. Medan kawat lurus (Hukum Ampere)      : B_phi = {B_wire}")

# 3. Potensial Vektor A untuk kawat lurus (Gauge Coulomb):
s0_s = sp.symbols('s_0', positive=True, real=True)
A_z = - (mu0_s * I_s / (2 * sp.pi)) * sp.log(s_s / s0_s)
B_phi_from_A = sp.simplify(- sp.diff(A_z, s_s))
display(sp.Eq(sp.Symbol(r'(\nabla \times \mathbf{A})_\phi'), B_phi_from_A))
print(f"4. Verifikasi curl(A) = B                : {sp.simplify(B_phi_from_A - B_wire) == 0} (Eksak identik)")
```

## 7. Rangkuman Komparatif Formula & Pembahasan Fisis

Hasil analisis komprehensif magnetostatika dirangkum dalam tabel komparatif berikut:

| Konsep / Fenomena | Persamaan Analitis Pokok | Arah Vektor & Simetri | Pangkat Peluruhan / Batas Jarak |
| :--- | :---: | :--- | :---: |
| **Gaya Lorentz pada Muatan** | $\mathbf{F} = q(\mathbf{E} + \mathbf{v}\times\mathbf{B})$ | Tegak lurus kecepatan $\mathbf{v}$ dan medan $\mathbf{B}$ (kerja magnet $W=0$) | Tidak bergantung jarak langsung |
| **Gaya pada Elemen Arus** | $d\mathbf{F} = I\,d\mathbf{l}\times\mathbf{B}$ | Mengikuti kaidah tangan kanan perkalian silang | Dua kawat sejajar: $F/L \propto 1/d$ |
| **Hukum Biot–Savart (Umum)** | $d\mathbf{B} = \dfrac{\mu_0 I}{4\pi}\dfrac{d\mathbf{l}'\times\hat{\boldsymbol{\mathcal{R}}}}{\mathcal{R}^2}$ | Tegak lurus terhadap elemen arus $d\mathbf{l}'$ dan vektor perpindahan $\hat{\boldsymbol{\mathcal{R}}}$ | Kuadrat terbalik jarak: $\mathcal{R}^{-2}$ |
| **Kawat Lurus Tak Hingga** | $\mathbf{B} = \dfrac{\mu_0 I}{2\pi s}\,\hat{\boldsymbol{\phi}}$ | Melingkar azimutal mengelilingi sumbu arus | Berbanding terbalik jarak: $s^{-1}$ |
| **Sumbu Loop Lingkaran ($R$)** | $B_z(z) = \dfrac{\mu_0 I R^2}{2(R^2+z^2)^{3/2}}$ | Sepanjang sumbu aksial simetri loop ($+z$) | Pusat ($z=0$): $\mu_0 I / (2R)$<br>Dipol ($z \gg R$): $z^{-3}$ |
| **Solenoida Panjang Ideal ($n$)** | $\mathbf{B} = \mu_0 n I\,\hat{\mathbf{z}}$ | Seragam di dalam rongga, nol di luar | Independen terhadap jari-jari dan posisi |
| **Toroida Ideal ($N$)** | $\mathbf{B} = \dfrac{\mu_0 N I}{2\pi s}\,\hat{\boldsymbol{\phi}}$ | Murni azimutal di dalam inti toroida, nol di luar | Berbanding terbalik terhadap radius $s^{-1}$ |
| **Divergensi Medan $\mathbf{B}$** | $\nabla \cdot \mathbf{B} = 0$ | Fluks magnetik netto permukaan tertutup selalu nol ($\oint \mathbf{B}\cdot d\mathbf{a} = 0$) | Tidak ada muatan/monopol magnetik |
| **Curl Medan $\mathbf{B}$ (Ampère)** | $\nabla \times \mathbf{B} = \mu_0 \mathbf{J}$ | Arus volumetrik lokal bertindak sebagai pusaran medan $\mathbf{B}$ | $\oint \mathbf{B}\cdot d\mathbf{l} = \mu_0 I_{\text{enc}}$ |
| **Potensial Vektor Magnet** | $\mathbf{B} = \nabla \times \mathbf{A}, \quad \nabla^2\mathbf{A} = -\mu_0\mathbf{J}$ | Memenuhi tara Coulomb $\nabla\cdot\mathbf{A}=0$ | $\mathbf{A}(\mathbf{r}) = \dfrac{\mu_0}{4\pi}\int \dfrac{\mathbf{J}(\mathbf{r}')}{\mathcal{R}}d\tau'$ |

### Prinsip Fisis Konseptual:
1. **Ketiadaan Monopol Magnetik**: Sifat solenoida $\nabla \cdot \mathbf{B} = 0$ berlaku universal dalam elektrodinamika klasik, menjamin bahwa garis-garis medan magnet tidak pernah berawal atau berakhir pada suatu titik, melainkan selalu membentuk loop tertutup atau membentang ke tak hingga.
2. **Ketiadaan Kerja Gaya Magnet**: Karena gaya magnetik $\mathbf{F}_{\text{mag}} = q(\mathbf{v} \times \mathbf{B})$ selalu tegak lurus kecepatan partikel ($\mathbf{F} \cdot \mathbf{v} = 0$), medan magnet tidak pernah melakukan usaha mekanik pada muatan partikel bebas ($dW/dt = 0$); medan magnet hanya membelokkan arah gerak lintasan tanpa mengubah energi kinetik.
3. **Resiprositas Induktansi Silang**: Hubungan fluks bersama antara dua loop arus tertutup $M_{12} = M_{21} = \frac{\mu_0}{4\pi}\oint\oint \frac{d\mathbf{l}_1 \cdot d\mathbf{l}_2}{\mathcal{R}_{12}}$ membuktikan simetri geometris fundamental hukum Neumann.

## 8. Latihan dan Pembahasan

Soal-soal berikut **disusun ulang sebagai latihan orisinal** untuk menerapkan gagasan Bab 5; ini bukan salinan redaksi soal Griffiths. Setiap solusi menunjukkan penentuan arah, proyeksi, dan pemeriksaan satuan. Rujukan di akhir soal menunjukkan bagian materi yang mendasarinya.

### Soal 1: Gaya Lorentz dan kerja magnet

Sebuah partikel bermuatan $q=-2.0\,\mu\mathrm C$ memiliki kecepatan $\mathbf v=(3,-4,2)\times10^5\,\mathrm{m/s}$. Pada titiknya terdapat $\mathbf B=0.50\,\hat{\mathbf z}\,\mathrm T$ dan $\mathbf E=2.0\times10^4\,\hat{\mathbf x}\,\mathrm{N/C}$. Tentukan gaya magnet dan gaya total dalam komponen Kartesius. Verifikasi bahwa gaya magnet tidak melakukan kerja sesaat.

**Solusi.** Hitung hasil silang sebelum mengalikan dengan muatan:

$$\mathbf v\times\mathbf B=(v_yB_z)\hat{\mathbf x}-(v_xB_z)\hat{\mathbf y}=(-2.0\times10^5\hat{\mathbf x}-1.5\times10^5\hat{\mathbf y})\,\mathrm{m\,T/s}.$$

Karena muatannya negatif, arah gaya berlawanan dengan hasil silang tersebut:

$$\mathbf F_{\rm mag}=q(\mathbf v\times\mathbf B)=(0.40\hat{\mathbf x}+0.30\hat{\mathbf y})\,\mathrm N.$$

Gaya listriknya $q\mathbf E=-0.040\hat{\mathbf x}\,\mathrm N$, sehingga $\mathbf F=(0.36\hat{\mathbf x}+0.30\hat{\mathbf y})\,\mathrm N$. Pemeriksaan kerja magnet:

$$\mathbf F_{\rm mag}\cdot\mathbf v=(0.40)(3.0\times10^5)+(0.30)(-4.0\times10^5)=0\,\mathrm W.$$

Komponen $v_z$ sejajar $\mathbf B$ dan tidak berkontribusi pada hasil silang. (Konsep: Griffiths §5.1.2, hlm. 212–216.)

### Soal 2: gaya pada segmen kawat dan proyeksi medan

Segmen kawat lurus sepanjang $0.25\,\mathrm m$ dialiri arus $4.0\,\mathrm A$ ke arah $+x$. Medan seragamnya $\mathbf B=(0.30\hat{\mathbf y}+0.40\hat{\mathbf z})\,\mathrm T$. Tentukan vektor gaya dan besarnya.

**Solusi.** Karena arus menuju $+x$, $d\boldsymbol\ell=0.25\hat{\mathbf x}\,\mathrm m$. Maka

$$\mathbf F=I\,d\boldsymbol\ell\times\mathbf B
=4.0(0.25)\left[0.30(\hat{\mathbf x}\times\hat{\mathbf y})+0.40(\hat{\mathbf x}\times\hat{\mathbf z})\right]$$
$$\hspace{1.5cm}=(-0.40\hat{\mathbf y}+0.30\hat{\mathbf z})\,\mathrm N.$$

Kedua komponen medan tegak lurus kawat, sehingga seluruh medan berkontribusi. Besar gaya $|\mathbf F|=\sqrt{0.40^2+0.30^2}=0.50\,\mathrm N$. Tanda negatif pada komponen $y$ berasal dari $\hat{\mathbf x}\times\hat{\mathbf z}=-\hat{\mathbf y}$. (Konsep: Griffiths §5.1.3, hlm. 216–220.)

### Soal 3: Medan kawat lurus dari Biot–Savart

Kawat lurus tak hingga berimpit dengan sumbu $z$ dan membawa arus $I=3.0\,\mathrm A$ ke arah $+z$. Cari medan di titik berjarak tegak lurus $s=6.0\,\mathrm{cm}$ pada sumbu $+x$. Tunjukkan proyeksi yang menentukan arah medan.

**Solusi.** Parametrisasikan elemen sumber sebagai $d\boldsymbol\ell'=dz'\hat{\mathbf z}$ dan vektor dari elemen itu ke titik pengamatan sebagai $\mathbf R=s\hat{\mathbf x}-z'\hat{\mathbf z}$. Perkalian silang memberi

$$d\boldsymbol\ell'\times\hat{\mathbf R}=\frac{s\,dz'}{\sqrt{s^2+z'^2}}\hat{\mathbf y};$$

komponen sepanjang $z$ pada $\mathbf R$ hilang karena $\hat{\mathbf z}\times\hat{\mathbf z}=0$. Semua elemen menyumbang ke arah $+y$. Besarnya:

$$B=\frac{\mu_0 I}{4\pi}\int_{-\infty}^{\infty}\frac{s\,dz'}{(s^2+z'^2)^{3/2}}=\frac{\mu_0 I}{2\pi s}.$$

Dengan $s=0.060\,\mathrm m$, diperoleh $\mathbf B=1.0\times10^{-5}\hat{\mathbf y}\,\mathrm T$. Arah $+y$ cocok dengan aturan tangan kanan untuk arus $+z$ pada titik $+x$. (Konsep: Griffiths §5.2.2, hlm. 224–227.)

### Soal 4: Hukum Ampère untuk kawat silinder pejal

Kawat silinder panjang berjari-jari $a=2.0\,\mathrm{cm}$ membawa arus total $I=8.0\,\mathrm A$ yang terdistribusi merata pada penampang. Cari medan pada $\rho=1.0\,\mathrm{cm}$ dan $\rho=8.0\,\mathrm{cm}$ dari sumbu.

**Solusi.** Simetri silinder membuat $\mathbf B=B(\rho)\hat{\boldsymbol\phi}$ dan besarannya konstan pada lingkaran Amperian berjari-jari $\rho$. Untuk $\rho<a$, rapat arus seragam memberi $I_{\rm enc}=I\rho^2/a^2$. Jadi

$$B(\rho<a)(2\pi\rho)=\mu_0I_{\rm enc},\qquad B(\rho<a)=\frac{\mu_0I\rho}{2\pi a^2}.$$

Pada $\rho=0.010\,\mathrm m$, $B=4.0\times10^{-5}\,\mathrm T$. Di luar kawat seluruh arus terlingkupi, sehingga $B(\rho>a)=\mu_0I/(2\pi\rho)$; pada $\rho=0.080\,\mathrm m$, $B=2.0\times10^{-5}\,\mathrm T$. Kedua arah adalah azimutal $+\hat{\boldsymbol\phi}$ untuk arus $+z$. Sebagai pemeriksaan, rumus dalam dan luar sama-sama memberi $B=\mu_0I/(2\pi a)$ tepat di permukaan kawat. (Konsep: Griffiths §5.3.3, hlm. 233–239.)

### Soal 5: Memperoleh medan dari potensial vektor

Untuk kawat lurus tak hingga berarus $I$ pada sumbu $z$, pertimbangkan pilihan gauge (untuk $s>0$)

$$\mathbf A=A_z(s)\hat{\mathbf z},\qquad A_z(s)=-\frac{\mu_0I}{2\pi}\ln\left(\frac{s}{s_0}\right),$$

dengan $s_0$ jarak acuan. Hitung $\nabla\times\mathbf A$, lalu jelaskan apa yang terjadi jika ditambahkan gradien $\nabla\chi$ dengan $\chi=c z$.

**Solusi.** Dalam koordinat silinder, jika hanya $A_z(s)$ yang tidak nol, komponen azimutal curl ialah $(\nabla\times\mathbf A)_\phi=-\partial A_z/\partial s$. Karena $dA_z/ds=-\mu_0I/(2\pi s)$,

$$\mathbf B=\frac{\mu_0I}{2\pi s}\hat{\boldsymbol\phi}.$$

Untuk $I=3.0\,\mathrm A$ dan $s=5.0\,\mathrm{cm}$, $B=1.2\times10^{-5}\,\mathrm T$. Sekarang $\nabla\chi=c\hat{\mathbf z}$ adalah vektor konstan, sehingga $\mathbf A'=\mathbf A+c\hat{\mathbf z}$ dan $\nabla\times\mathbf A'=\nabla\times\mathbf A$; medan fisik tidak berubah. Fungsi $s_0$ juga hanya mengubah $A_z$ dengan konstanta, bukan $\mathbf B$. (Konsep: Griffiths §5.4.1, hlm. 243–249.)
